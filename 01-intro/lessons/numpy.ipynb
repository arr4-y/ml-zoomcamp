```python
#-> crear arreglos
np.zeros(5)        # [0, 0, 0, 0, 0]
np.ones(5)         # [1, 1, 1, 1, 1]
np.full(5, 3.5)    # [3.5, 3.5, 3.5, 3.5, 3.5]
np.arange(0, 5)    # [0, 1, 2, 3, 4]
np.array([1,2,3])  # tu propia lista → array

#-> dimensiones (matrices)
np.zeros((3, 2))   # 3 filas, 2 columnas de ceros
n[0, 1]            # fila 0, columna 1
n[2]               # fila completa
n[:, 1]            # columna completa

#-> operaciones elemento por elemento:
a = np.array([1, 2, 3])
a + 10    # [11, 12, 13]
a * 2     # [2, 4, 6]
a ** 2    # [1, 4, 9]

#y entre dos arreglos también:
b = np.array([10, 20, 30])
a + b     # [11, 22, 33]

#-> comparaciones
a >= 2          # [False, True, True]
a[a >= 2]       # [2, 3] solo los que cumplen

#-> resumir datos
a.min()    # 1
a.max()    # 3
a.sum()    # 6
a.mean()   # 2.0

#-> números random
np.random.seed(42)              # para resultados reproducibles
np.random.rand(3, 2)            # random entre 0 y 1
np.random.randint(0, 100, 5)    # enteros random entre 0 y 99
```

```python
my_list = [2, 3, 4]
array_from_list = np.array(my_list)
array_from_list
```

Array Operations


Element-wise Operations


```python
arr = arr + 1   # Adds 1 to each element
arr = arr * 2   # Multiplies each element by 2
# Similar operations for division and exponentiation
```

```python
arr
```

Element-wise Operations with Two Arrays


```python
arr1 = np.ones(4)
arr2 = np.full(4, 3)
result = arr1 + arr2  # Element-wise addition
result = arr1 / arr2  # Element-wise division
```

```python
arr1
```

```python
arr2
```

```python
result
```

Comparison Operations


```python
arr = np.array([1, 2, 3, 4])
greater_than_2 = arr > 2  # Produces [False, False, True, True]
```

```python
arr
```

```python
greater_than_2
```

Selecting Elements Based on Conditions


```python
selected_elements = arr[arr > 1]  # Gets elements greater than 1
selected_elements
```

Summary Operations


```python
min_value = arr.min()    # Minimum value
max_value = arr.max()    # Maximum value
sum_value = arr.sum()    # Sum of all elements
mean_value = arr.mean()  # Mean (average) value
std_deviation = arr.std()  # Standard deviation
```
